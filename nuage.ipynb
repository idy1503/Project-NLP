```python
from wordcloud import WordCloud
import matplotlib.pyplot as mat
import nltk
import os
import pandas as pd
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
```

# Télechargement des librairies NLTK 

```python
nltk.download("punkt")
nltk.download("stopwords")
nltk.download("punkt_tab")
nltk.download("wordnet")
nltk.download("omw-1.4")
```

### Fonction de traitement des dse mots

```python
def processing(files):
    
    tokens = word_tokenize(files.lower())
    filte = [token for token in tokens if token not in stopwords.words('english') and token.isalpha()]

    lemmatiser = WordNetLemmatizer()

    lemmitazer = [lemmatiser.lemmatize(token) for token in filte]
    processed = ' '.join(lemmitazer)
    return processed
```

## Boucle d'affichage des nuages

```python
for i in os.listdir("Text/"):
    with open(f'Text/{i}', 'r', encoding='utf-8') as file:
        files = file.read()
        name = i.replace('.txt', '')
        l = processing(files)
        display(name)
        group = WordCloud(max_font_size=500, max_words=200, background_color="white", height=1500, width= 1500).generate(l)
        mat.imshow(group, interpolation="bilinear")
        mat.axis("off")
        mat.show()
```
