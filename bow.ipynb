# Imports

```python
from wordcloud import WordCloud
import matplotlib.pyplot as mat
import nltk
import os
import re
import pandas as pd
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
from collections import Counter
```

# Télechargement des librairies NLTK 

```python
nltk.download("punkt")
nltk.download("stopwords")
nltk.download("punkt_tab")
nltk.download("wordnet")
nltk.download("omw-1.4")
```

### Fonction de traitement des dse mots

```python
def bow(files):
    
    tokens = word_tokenize(files.lower())
    filte = [token for token in tokens if token not in stopwords.words('english') and token.isalpha()]

    lemmatiser = WordNetLemmatizer()

    lemmitazer = [lemmatiser.lemmatize(token) for token in filte]
    processed = Counter(lemmitazer)
    word_freq_df = pd.DataFrame(processed.items(), columns=['Word', 'Frequency'])
    frequency = word_freq_df.sort_values(by='Frequency', ascending=False)
    
    return frequency
```

### Boucle d'affichage des Bow

```python
for i in os.listdir("Text/"):
    with open(f'Text/{i}', 'r', encoding='utf-8') as file:
        files = file.read()
        name = i.replace('.txt', '')
        l = bow(files)
        display(name)
        display(l)
        
```
